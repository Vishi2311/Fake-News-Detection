```python
import pandas as pd

fake = pd.read_csv("../data/raw/Fake.csv")
true = pd.read_csv("../data/raw/True.csv")

print("Fake news:", fake.shape)
print("Real news:", true.shape)
```

```python
print("Fake dataset columns:")
print(fake.columns.tolist())

print("\nReal dataset columns:")
print(true.columns.tolist())
```

```python
fake.head()
```

```python
true.head()
```

```python
print("Fake dataset missing values:")
print(fake.isnull().sum())

print("\nReal dataset missing values:")
print(true.isnull().sum())
```

```python
fake["label"] = 0
true["label"] = 1

df = pd.concat([fake, true], ignore_index=True)

print("Total articles:", len(df))
print("\nLabel counts:")
print(df["label"].value_counts())
```

```python
df = df.sample(frac=1, random_state=42).reset_index(drop=True)

print(df.head())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates().reset_index(drop=True)

print("Rows after removing duplicates:", len(df))
```

```python
df.to_csv("../data/processed/cleaned_news.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python
print("Dataset shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nLabel distribution:")
print(df["label"].value_counts())

print("\nSample article:")
print(df[["title", "text", "label"]].head(2))
```
